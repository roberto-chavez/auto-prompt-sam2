```
import hashlib, json, shutil
from pathlib import Path

# Definir la ruta de destino en tu Drive para guardar los resultados automáticamente
drive_output_dir = Path('/content/drive/MyDrive/auto-prompt-sam2/outputs')
drive_output_dir.mkdir(parents=True, exist_ok=True)

# Buscar el chip en Google Drive
project = Path('/content/drive/MyDrive/auto-prompt-sam2')
relative = Path('raw/v1.1/data/flood_events/HandLabeled/S2Hand/Ghana_1033830_S2Hand.tif')
candidates = (project / relative, project / 'auto-prompt-sam2' / relative)
source = next((path for path in candidates if path.is_file()), None)

if source is None:
    raise FileNotFoundError(f'No se encontró el archivo en Drive. Asegúrate de colocarlo en: {project / relative}')

chip = Path('/content/chip.tif')
shutil.copyfile(source, chip)
print('sha256 chip:', hashlib.sha256(chip.read_bytes()).hexdigest())
```

```
from auto_prompt_sam2.demo_config import load_demo_config
from auto_prompt_sam2.demo_backend import DemoBackend
from auto_prompt_sam2.demo_images import load_image, S2_LAYOUT
cfg=load_demo_config(environ={**os.environ,'DEMO_MODE':'inference','DEMO_ALLOW_DOWNLOAD':'1'})
be=DemoBackend(cfg)
img=load_image('/content/chip.tif',S2_LAYOUT,be.params)
print(img.rgb.shape, 'válidos', int(img.valid.sum()))
# Puntos manuales (x,y,etiqueta): ajustar tras mirar img.preview(); 1=agua, 0=fondo. Deben caer en píxeles válidos.
import matplotlib.pyplot as plt; plt.imshow(img.preview()); plt.show()
```

```
from PIL import Image

POINTS = [(256, 256, 1), (100, 100, 0)]  # Ajusta tus puntos según convenga
out = be.segment(img, POINTS, compare=False)
m = out['frozen']['mask']
print('píxeles de máscara:', int(m.sum()), 'estado:', out['status'])

assert m.dtype == bool and m.shape == img.valid.shape and m.sum() > 0, 'La máscara salió vacía'

# Guardar la imagen y el JSON directamente en Google Drive
Image.fromarray((m * 255).astype('uint8')).save(drive_output_dir / 'e3_demo_mask.png')

json.dump({
    'points': POINTS,
    'mask_pixels': int(m.sum()),
    'selection': str(out['frozen']['selection']),
    'gpu': torch.cuda.get_device_name(0),
    'chip_sha256': hashlib.sha256(open('/content/chip.tif', 'rb').read()).hexdigest(),
    'checkpoint': 'facebook/sam2.1-hiera-tiny'
}, open(drive_output_dir / 'e3_demo_smoke.json', 'w'), indent=1)

print('OK e3_demo_smoke guardado de forma segura en Google Drive.')
```
